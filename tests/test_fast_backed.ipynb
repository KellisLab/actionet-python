```python
import actionet
import anndata
import pandas as pd
import numpy as np

import lets_plot
lets_plot.LetsPlot.setup_html(no_js=True)
```

```python
adata_src = anndata.read_h5ad("../data/test_adata.h5ad", backed = "r")
adata_src.copy(filename="../data/adata_working_test.h5ad")
adata_src.file.close()
```

```python
adata = anndata.read_h5ad("../data/adata_working_test.h5ad", backed = "r+")
```

```python
actionet.filter_anndata(adata, min_cells_per_feat=0.01, inplace=True)
actionet.normalize_anndata(adata, target_sum = 1e4, log_base=2, log_transform=True, inplace=True, dtype_out="float32")
```

```python
actionet.reduce_kernel(adata, n_components=30, key_added='action', inplace=True, backed_n_threads=8, svd_algorithm="auto", max_iter=0)
```

```python
# actionet.correct_batch_effect(adata, batch_key="UID", inplace=True)
```

```python
actionet.run_actionet(adata, inplace=True, reduction_key="action")
```

```python
actionet.plot_umap(adata, color='CellLabel', basis='umap_2d_actionet')
```

```python
actionet.plot_feature_expression(adata, features=["Drd1", "Drd2", "Mbp", "Gpc5", "Vcan", "Adarb2"], features_use="Gene", single_plot=True, layer=None)
```

```python
markers = actionet.find_markers(adata, adata.obs['CellLabel'], features_use="Gene", top_genes=30, return_type='dataframe')
markers
```

```python
annots_out = actionet.annotate_cells(adata, markers, layer = None, method='vision', features_use='Gene', use_enrichment=True, use_lpa=False, n_threads=0)
```

```python
actionet.plot_umap(adata, color=annots_out['labels'], basis='umap_2d_actionet')
```
